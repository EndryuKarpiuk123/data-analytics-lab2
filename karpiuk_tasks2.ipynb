## Завдання 2. Завантаження `sales.csv`, `head(3)` та `tail(3)`

```python
from google.colab import drive
drive.mount('/content/drive')
```

```python
PATH = '/content/drive/MyDrive/sales.csv'

# Якщо файлу sales.csv немає на Диску - створюємо навчальний набір про продажі кави
import os
if not os.path.exists(PATH):
    rng = np.random.default_rng(42)
    n = 60
    drinks = {"Latte": 38, "Americano": 30, "Cappuccino": 35, "Espresso": 28, "Hot Chocolate": 36}
    names = rng.choice(list(drinks), n)
    cash = rng.choice(["card", "cash"], n, p=[0.8, 0.2])
    dt = pd.to_datetime("2024-03-01 08:00") + pd.to_timedelta(np.sort(rng.integers(0, 60*24*20, n)), unit="m")
    pd.DataFrame({
        "date": dt.date,
        "datetime": dt,
        "cash_type": cash,
        "card": [f"ANON-{rng.integers(1000,9999)}" if c == "card" else np.nan for c in cash],
        "money": [drinks[x] + rng.choice([0, 2, 4]) for x in names],
        "coffee_name": names,
    }).to_csv(PATH, index=False)
    print("Створено навчальний файл:", PATH)

df = pd.read_csv(PATH)
print('Перші 3 рядки:')
display(df.head(3))
print('Останні 3 рядки:')
display(df.tail(3))
```

## ★ Додаткове завдання: функція підрахунку записів за категорією


```python
def count_by_category(frame, column):
    """Повертає DataFrame з кількістю записів у кожній групі колонки column (за спаданням)."""
    counts = frame[column].value_counts().reset_index()   # кількість записів у групах
    counts.columns = [column, 'count']
    return counts.sort_values('count', ascending=False)   # сортування за спаданням

result = count_by_category(df, 'coffee_name')
result
```
