## Крок 3. Імпорт даних з CSV-файлу

```python
from google.colab import drive  # модуль для роботи з Google Диском
drive.mount('/content/drive')    # монтує Диск
```

```python
PATH = '/content/drive/MyDrive/sales.csv'

# Якщо файлу sales.csv немає на Диску - створюємо навчальний набір про продажі кави
import os
if not os.path.exists(PATH):
    rng = np.random.default_rng(42)
    n = 60
    drinks = {"Latte": 38, "Americano": 30, "Cappuccino": 35, "Espresso": 28, "Hot Chocolate": 36}
    names = rng.choice(list(drinks), n)
    cash = rng.choice(["card", "cash"], n, p=[0.8, 0.2])
    dt = pd.to_datetime("2024-03-01 08:00") + pd.to_timedelta(np.sort(rng.integers(0, 60*24*20, n)), unit="m")
    pd.DataFrame({
        "date": dt.date,
        "datetime": dt,
        "cash_type": cash,
        "card": [f"ANON-{rng.integers(1000,9999)}" if c == "card" else np.nan for c in cash],
        "money": [drinks[x] + rng.choice([0, 2, 4]) for x in names],
        "coffee_name": names,
    }).to_csv(PATH, index=False)
    print("Створено навчальний файл:", PATH)

df_sales = pd.read_csv(PATH)  # зчитує CSV у DataFrame
df_sales.head()               # перші п'ять рядків
```

## Крок 4. Перегляд даних

```python
df_sales.describe()  # статистики для числових колонок
```

## Крок 5. Індексація за допомогою `loc` та `iloc`

```python
# Третій рядок за числовою позицією
row_iloc = df_sales.iloc[2]
print(row_iloc)
```

```python
# Рядок з міткою індексу 2
row_loc = df_sales.loc[2]
print(row_loc)
```

```python
# Значення колонки 'money' для рядка з індексом 5
value = df_sales.loc[5, 'money']
print(value)
```

## Крок 6. Фільтрація даних за умовою


```python
mask_high = df_sales['money'] > 35   # булева маска (True/False для кожного рядка)
high_sales = df_sales[mask_high]     # лише рядки, де маска True
print(high_sales.head())
```

## Крок 7. Сортування даних

```python
sorted_sales = df_sales.sort_values('money', ascending=False)  # за спаданням суми
print(sorted_sales.head())
```

```python
# Різниця між iloc та loc після сортування
print("iloc[0] - перший рядок у новому порядку:")
print(sorted_sales.iloc[0])
print()
print("loc[0] - рядок з оригінальною міткою індексу 0:")
print(sorted_sales.loc[0])
```

## Крок 8. Збереження результатів


```python
final = df_sales[df_sales['money'] > 35].sort_values('money', ascending=False)

final.to_csv('/content/drive/MyDrive/filtered_sales.csv', index=False)  # index=False - без колонки індексів
print("Файл збережено!")
```
